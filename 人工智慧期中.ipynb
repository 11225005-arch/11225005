```python
# 步驟十一：載入 MNIST 手寫數字資料集

import tensorflow as tf
from tensorflow.keras.datasets import mnist

# 載入訓練資料與測試資料
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# 顯示資料集資訊
print("TensorFlow 版本：", tf.__version__)
print("訓練資料筆數：", len(x_train))
print("測試資料筆數：", len(x_test))
print("每張圖片尺寸：", x_train[0].shape)
print("第一張圖片的正確答案：", y_train[0])
```

```python
# 步驟十二：顯示 MNIST 手寫數字圖片

import matplotlib.pyplot as plt

# 建立圖片畫布
plt.figure(figsize=(10, 5))

# 顯示前 10 張訓練圖片
for i in range(10):
    plt.subplot(2, 5, i + 1)

    # 顯示灰階圖片
    plt.imshow(x_train[i], cmap='gray')

    # 顯示正確的數字標籤
    plt.title(f"Label: {y_train[i]}")

    # 隱藏座標軸
    plt.axis('off')

plt.suptitle("MNIST Handwritten Digits", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# 步驟十三：MNIST 資料前處理

# 將像素值從 0～255 轉換成 0～1
x_train_normalized = x_train.astype("float32") / 255.0
x_test_normalized = x_test.astype("float32") / 255.0

# 顯示前處理結果
print("訓練資料形狀：", x_train_normalized.shape)
print("測試資料形狀：", x_test_normalized.shape)

print("正規化前最大像素值：", x_train.max())
print("正規化後最大像素值：", x_train_normalized.max())

print("正規化後最小像素值：", x_train_normalized.min())
```

```python
# 步驟十四：建立 MNIST 神經網路模型

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Flatten, Dense

# 建立神經網路
model = Sequential([
    Input(shape=(28, 28)),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

# 設定模型訓練方式
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 顯示模型架構
model.summary()
```

```python
# 步驟十五：訓練 MNIST 神經網路模型

# 開始訓練模型
history = model.fit(
    x_train_normalized,
    y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)

# 顯示訓練完成訊息
print("MNIST 神經網路模型訓練完成！")
```

```python
# 步驟十六：MNIST 模型最終測試

# 使用 10,000 張測試圖片評估模型
test_loss, test_accuracy = model.evaluate(
    x_test_normalized,
    y_test,
    verbose=1
)

# 顯示測試結果
print("\n===== MNIST 模型測試結果 =====")
print(f"測試資料筆數：{len(x_test_normalized)}")
print(f"測試準確率：{test_accuracy * 100:.2f}%")
print(f"測試損失：{test_loss:.4f}")
```

```python
# 步驟十七：繪製 MNIST 模型訓練曲線

import matplotlib.pyplot as plt

# 建立圖表
plt.figure(figsize=(12, 5))

# 左圖：準確率
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], marker='o',
         label='Training Accuracy')
plt.plot(history.history['val_accuracy'], marker='o',
         label='Validation Accuracy')

plt.title('MNIST Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

# 右圖：損失值
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], marker='o',
         label='Training Loss')
plt.plot(history.history['val_loss'], marker='o',
         label='Validation Loss')

plt.title('MNIST Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# 步驟十八：MNIST AI 手寫數字預測結果

import numpy as np
import matplotlib.pyplot as plt

# 使用測試資料的前 10 張圖片進行預測
predictions = model.predict(x_test_normalized[:10])

# 取得預測數字
predicted_labels = np.argmax(predictions, axis=1)

# 建立圖片畫布
plt.figure(figsize=(12, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)

    # 顯示手寫數字圖片
    plt.imshow(x_test[i], cmap='gray')

    # 比較 AI 預測與正確答案
    predicted = predicted_labels[i]
    actual = y_test[i]

    # 正確顯示綠色，錯誤顯示紅色
    color = 'green' if predicted == actual else 'red'

    plt.title(
        f"Pred: {predicted} | True: {actual}",
        color=color
    )

    plt.axis('off')

plt.suptitle("MNIST AI Prediction Results", fontsize=16)
plt.tight_layout()
plt.show()
```
